# Generator
Generator, generate a graph with n nodes. edges matrix is an n*n matrix that edges[i][j] is the weight of the edge between nodes i and j.

```python
def Generator(n, seed=42) :
    random.seed(seed)
    edges = np.zeros((n,n))
    for i in range(n) :
        for j in range(n) :
            if i >= j :
                continue
            edges[i][j] = random.randint(0,100)
            edges[j][i] = edges[i][j]
    return edges
```

# Brute force

```python
def brute_force(n, k, edges,picked=set()) :
    '''
        input
            n = number of nodes
            k = number of nodes to pick
            picked = already picked nodes
        output
            ans = best answer that contains the nodes of 'picked'
    '''

    # picked_list= list(range(1,k+1))
    # best_list=picked_list
    # best=current=get_edge_sum(edges, picked_list)

    # while(True):
    #   print(picked_list)
    #   picked_list=get_next_iteration(n, picked_list)
    #   if(picked_list == None):
    #     break
    #   current=get_edge_sum(edges, picked_list)

    #   if(current> best):
    #     best_list=picked_list
    #     best=current
    # return best


    n = edges.shape[0]
    if len(picked) == k :
        ans = 0
        for u in picked :
            for v in picked :
                if v > u :
                    ans += edges[u][v]
        return ans

    ans = 0
    for v in range(n) :
        if v not in picked :
             picked.add(v)
             ans = max(ans, brute_force(n, k, edges, picked))
             picked.remove(v)
    return ans
```

# Hill climbing

```python
def random_choice(n, k):
    '''
        choose k random unique number between 1 to n
    '''
    output = []

    output=random.sample(range(n) ,k)

    return output
```

```python
def get_value(state, edges):
    '''
        state is a list that contains some nodes
        return sum of edges of state nodes
    '''
    ans=0
    for vertice1 in state:
      for vertice2 in state:
        if(vertice2 == vertice1 or vertice2 < vertice1):
          continue
        ans +=edges[vertice1][vertice2]
    return ans
```

We consider two states like X and Y as neighbor states, if there is exactly one node in X that isn't in Y and there is exactly one node in Y that isn't in X.

For instance [2,3,4,5,6] and [2,3,4,6,7] are neighbors but [2,3,4,5,6] and [1,2,3,7,8] aren't neighbors.

```python
def get_neighbors(n, state):
    '''
        return neighbors of state
    '''
    neighbors = []
    for i in state:
      temp=state[:]
      temp.remove(i)
      for j in range(n):
        if(j not in state):
          temp.append(j)
          temp.sort()
          neighbors.append(temp[:])
          temp.remove(j)
    return neighbors
```

```python
def hill_climbing(n, k, edges, num_iters = 10000):
    '''
        input
            n = number of graph nodes
            k = number of nodes to pick
            edges = graph edges weights
            num_iters = maximum number of iterations
        output
            best_value = best state value
    '''
    best_value=0
    current_state = random_choice(n,k)
    current_value = get_value(current_state, edges)

    while num_iters > 0 :
        num_iters -= 1
        neighbors=get_neighbors(n,current_state)
        best_neighbor=[]
        best_neighbor_value=0
        for newstate in neighbors:
          newvalue=get_value(newstate, edges)
          if(newvalue > best_neighbor_value):
            best_neighbor_value=newvalue
            best_neighbor=newstate

        if(current_value < best_neighbor_value):
          current_state=best_neighbor
          current_value=best_neighbor_value
        else:
          if(current_value > best_value):
            best_value=current_value
          current_state = random_choice(n,k) #random restart to check answer with another iteration to gurantee reaching the optimal answer
          current_value = get_value(current_state, edges)


    return best_value
```

```python
ans = hill_climbing(n, k, edges)
print(ans)
```

# Simulated annealing

```python
def bernoulli_trial(p):
    return 1 if random.random() < p else 0

def simulated_annealing(n, k, edges, alpha = 0.9, initial_temp = 100000, max_iteration = 100000, min_temperature = 0.0001):
    '''
    input
        alpha is the temperature decay rate
        max_iteration is the maximum number of iteration (termination condition)
        min_temperature is the minimum temperature (termination condition)
    output
        best state value
    '''
    temperature = initial_temp
    best_value = float('-INF')


    current_state = random_choice(n, k)
    while(temperature >= min_temperature and max_iteration > 0):
        max_iteration -= 1
        nextState = random.choice( get_neighbors(n, current_state))
        next_value=get_value(nextState, edges)

        delta=next_value-best_value
        if (delta>0):
          best_value=next_value
          current_state= nextState
        else:
          exp_value = math.exp(delta / temperature) if delta / temperature > -700 else 0  # Clamp to avoid underflow
          if(bernoulli_trial(exp_value)): #only perform bad move if trial is won
            best_value=next_value
            current_state= nextState
        temperature = alpha * temperature
    return best_value
```

```python
ans = simulated_annealing(n, k, edges)
print(ans)
```
