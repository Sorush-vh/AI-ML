```python
def is_empty(sudoku):
    """
    TODO: This function should check whether the sudoku is still empty (contains zero values)
    """
    for i in range(9):
        for j in range(9):
            if sudoku[i][j]==0:
                return True
    return False
```

```python
def is_placement_valid(grid, num, position):
    """
    TODO: This function determines whether it's valid to place the number 'num' at the given 'position' on the Sudoku grid.
    """
    row, col = position
    #Check if the number is already in the row
    for i in range(9):
        if grid[row][i] == num:
            return False
    #Check if number is already in the col:
    for i in range(9):
        if grid[i][col] == num:
            return False

    #Check if number is already in the block:
    block_base_row = row-(row%3)
    block_base_col = col-(col%3)
    for i in range(3):
        for j in range(3):
            if grid[i + block_base_row][j + block_base_col] == num:
                return False
    return True
```

```python
def get_an_empty_position(soduku):
  for i in range(9):
    for j in range(9):
      if soduku[i][j]==0:
        return (i,j)


def backtrack(sudoku):
    """
    TODO: implement the backtracking algorithm utilizing is_empty and is_valid functions
    """
    if not is_empty(sudoku):
        return True
    
    row,col=get_an_empty_position(sudoku)
    
    for i in range(1,10):
      if is_placement_valid(sudoku,i,(row,col)):

        sudoku[row][col]=i
        if backtrack(sudoku):
           return True
        sudoku[row][col]=0
    return False
```

```python
from ortools.sat.python import cp_model
import pandas as pd
import numpy as np
```

```python

def solve_sudoku(sudoku):
    """
    Solves the given Sudoku puzzle using OR-Tools and returns the solved grid.

    :param sudoku: 2D numpy array representing the Sudoku puzzle (0 for empty cells)
    :return: Solved 2D numpy array or a message indicating failure
    """
    model = cp_model.CpModel()
    n = 9  # Standard Sudoku is 9x9

    # Create the variables for each cell in the Sudoku grid
    cells = [[model.NewIntVar(1, 9, f'cell_{i}_{j}') for j in range(n)] for i in range(n)]

    # Add constraints for rows, columns, and 3x3 subgrids
    for i in range(n):
        model.AddAllDifferent(cells[i])  # Row constraint
        model.AddAllDifferent([cells[j][i] for j in range(n)])  # Column constraint

    # 3x3 subgrid constraints
    for row in range(0, n, 3):
        for col in range(0, n, 3):
            model.AddAllDifferent([cells[row + i][col + j] for i in range(3) for j in range(3)])

    # Set the initial known values in the Sudoku puzzle
    for i in range(n):
        for j in range(n):
            if sudoku[i][j] != 0:
                model.Add(cells[i][j] == sudoku[i][j])

    # Solve the model
    solver = cp_model.CpSolver()
    status = solver.Solve(model)

    # Extract solution or return failure message
    if status == cp_model.FEASIBLE or status == cp_model.OPTIMAL:
        solved_grid = np.array([[solver.Value(cells[i][j]) for j in range(n)] for i in range(n)])
        return solved_grid
    else:
        return "No solution found for the given Sudoku puzzle."
```

### AC-3 Algorithm (20 Points)
First and foremost, implement the AC-3 algorithm to be later used in the problem

```python
from collections import deque

def remove_inconsistent_values(node1, node2, csp):
  removed=False
  for x in csp['domains'][node1]:
    satisfied=False
    for y in csp['domains'][node2]:
      if(x,y) in csp['constraints'][(node1,node2)]:
        satisfied=True
        break
    if(not satisfied):
      csp['domains'][node1].remove(x)
      removed=True
  return removed


def AC3(csp, neighbors):
    """
    csp:
        dict{
            nodes: list of nodes,
            domains: dictionary {variable: [domain]},
            constraints: dictionary {(node_i, node_j): list of constraints}
        },
    neighbors:
        dict {var: [neighbor]}

    TODO: Implement the AC3 algorithm

    return: a bool to show whether the problem is consistent and if so, the reduced domain of the problem
    """
    #initialize queue
    queue=deque()
    for node1 in csp['nodes']:
      for node2 in neighbors[node1]:
        if(csp['constraints'][(node1, node2)]): #if 2 neighbors have constraints on eachother
          queue.append((node1, node2))

    while(queue):
      node1, node2=queue.popleft()
      if(remove_inconsistent_values(node1 , node2, csp)):
        for neighbor in neighbors[node1]:
          queue.append((neighbor, node1))

    consistency=True
    for node in csp['nodes']:
      if(len(csp['domains'][node])==0):
        return False, None
    return consistency, csp['domains']
```

### AC-4 Algorithm (20 Points)
Secondly, implement the AC-4 (Arc Consistency 4) algorithm. This algorithm enforces arc consistency by maintaining **support tables** and efficiently processing **a queue of arcs** for consistency checks. To help you implement it, here are the key steps and concepts to focus on:
1. **Support Tables**: For each pair of variable-value assignments, you must maintain a table that tracks the number of supporting values in the neighboring variables' domains. This table will allow you to quickly determine if a value is consistent with its neighbors
2. Start by initializing your support tables for all variable-value pairs. For each arc (X, Y), compute and store the number of values in Y's domain that support each value in X's domain based on the constraint between X and Y.
3. Initialize **a queue of arcs** with all pairs of variables that share a constraint
4. For each arc (X, Y) in the queue, check the support for each value in X's domain. If any value in X loses all supporting values in Y, remove that value from X’s domain.
5. If a value is removed from X’s domain, for every neighboring variable Z (other than Y), you must add the arc (Z, X) back into the queue for further consistency checks. This ensures that removing a value from X doesn’t create inconsistencies for other variables.
6. Whenever a value is removed from a domain, update the support tables for all neighboring variable-value pairs that depend on the removed value. This way, if a value of Z is only supported by the value that was removed from X, the support count for that value of Z should be decremented.
7. The algorithm terminates when the queue is empty, meaning all arcs have been processed and the domains of all variables are arc-consistent. If at any point a domain becomes empty, the problem is unsolvable.

```python
def is_value_supported(csp, node , neighbor, value, neighbor_val):#based on constraints
  if((value, neighbor_val) in csp['constraints'][(node, neighbor)]):
    return True
  return False

def AC4(csp, neighbors):
    """
    csp:
        dict{
            nodes: list of nodes,
            domains: dictionary {variable: [domain]},
            constraints: dictionary {(node_i, node_j): list of constraints
        },

    TODO: Implement the AC4 algorithm

    return: a bool to show whether the problem is consistent and if so, the reduced domain of the problem
    """
    #initialize support tables
    support_dict={}#explicitly stores all values of a neighbor domain supporting a (var, val) pair
    support_table={}#storing the #of values of a neighbor domain supporting a (var, val) pair
    queue=deque()

    for node1 in csp['nodes']:
      for value in csp['domains'][node1]: #analyzing each (variable, value) pair:

        support_dict[(node1, value)]={}
        for neighbor in neighbors[node1]:#for all neighbors:

          support_dict[(node1, value)][neighbor]=[]
          support_table[(node1, value, neighbor)]=0

          for neighbor_val in csp['domains'][neighbor]:
            if(is_value_supported(csp, node1, neighbor, value, neighbor_val)):
              support_dict[(node1, value)][neighbor].append(neighbor_val)
              support_table[(node1, value, neighbor)]+=1
          if(support_table[(node1, value, neighbor)] ==0):
            queue.append((node1,value))

    while(queue):
      var, val =queue.popleft()
      csp['domains'][var].remove(val)
      if(not csp['domains'][var]):
        return False, None
      for neighbor in neighbors[var]:
        for neighbor_val in csp['domains'][neighbor]:
          if(val in support_dict[(neighbor, neighbor_val, var)]):
            support_table[(neighbor, neighbor_val, var)]-=1
            support_dict[(neighbor, neighbor_val, var)].remove(val)
            if(support_table[(neighbor, neighbor_val, var)]==0):
              queue.append(neighbor, neighbor_val)
    return True, csp['domains']
```
